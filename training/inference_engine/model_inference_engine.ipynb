```python
from tensor_rt_inference import BrainTumorClassifier
from PIL import Image
import time
```

```python
# --- Total time start ---
overall_start_time = time.time()

# --- Model loading ---
model_load_start = time.time()
model = BrainTumorClassifier("./brain_tumor.engine")
model_load_end = time.time()

# --- Input preparation ---
input_prep_start = time.time()
path = "../brain_tumor_dataset/Testing/no_tumor/image.jpg"
image = Image.open(path).convert("RGB")
input_prep_end = time.time()

# --- Inference ---
inference_start = time.time()
output = model.inference(image)
inference_end = time.time()

# --- Total time end ---
overall_end_time = time.time()

# --- Print timings in ms ---
print(f"Model Loading Time: {(model_load_end - model_load_start) * 1000:.2f} ms")
print(f"Input Preparation Time: {(input_prep_end - input_prep_start) * 1000:.2f} ms")
print(f"Inference Time: {(inference_end - inference_start) * 1000:.2f} ms")
print(f"Total Time (including setup): {(overall_end_time - overall_start_time) * 1000:.2f} ms")
```

```python
predicted_class = output.argmax().item()

print(predicted_class)  # Output: 1

# Mapping to label (optional):
class_mapping = {
    0: 'glioma_tumor',
    1: 'meningioma_tumor',
    2: 'no_tumor',
    3: 'pituitary_tumor'
}

print(class_mapping[predicted_class])  # Output: meningioma_tumor
```
