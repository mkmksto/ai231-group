```python
from utils import TumorClassifier, load_model_weights
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights
import cv2
import torch
from sklearn.preprocessing import LabelEncoder
```

```python
model = TumorClassifier(
    num_classes=4,
    pretrained_model=efficientnet_b0(weights=EfficientNet_B0_Weights.IMAGENET1K_V1),
) 
model = load_model_weights(model, "training_data/effnet_weights.pth")
```

```python
path = "brain_tumor_dataset/Testing/meningioma_tumor/image.jpg"

image_size = 150
img = cv2.imread(path)  # Convert Path object to string
img = cv2.resize(img, (image_size, image_size))

image = torch.FloatTensor(img) / 255.0 
image = image.permute(2, 0, 1)  # Change from HWC to CHW format
image = image.unsqueeze(0)      # (1, C, H, W)
```

```python
model.eval()
output = model(image)
```

```python
predicted_class = output.argmax(dim=1).item()

print(predicted_class)  # Output: 1

# Mapping to label (optional):
class_mapping = {
    0: 'glioma_tumor',
    1: 'meningioma_tumor',
    2: 'no_tumor',
    3: 'pituitary_tumor'
}

print(class_mapping[predicted_class])  # Output: meningioma_tumor
```
