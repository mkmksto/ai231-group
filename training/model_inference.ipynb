```python
from modules.model import TumorClassifier
from torchvision.models import efficientnet_b0, EfficientNet_B0_Weights
import cv2
import torch
from sklearn.preprocessing import LabelEncoder
from modules.transforms import val_transform
from PIL import Image
import time
```

```python
# Start overall timer
overall_start_time = time.time()

# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load model
model_load_start = time.time()
model = TumorClassifier()
model = model.to(device)
model.load_state_dict(torch.load("./training_data/effnet_weights_sampler_cosine_best_model.pth"))
model.eval()
model_load_end = time.time()

# Prepare input
input_prep_start = time.time()
path = "brain_tumor_dataset/Testing/meningioma_tumor/image.jpg"
transform = val_transform()

image = Image.open(path).convert("RGB")
transformed_image = transform(image)
input_tensor = transformed_image.unsqueeze(0).to(device)
input_prep_end = time.time()

# Inference
inference_start = time.time()
model.eval()
output = model(input_tensor)
inference_end = time.time()

# End overall timer
overall_end_time = time.time()

# Print timings in milliseconds
print(f"Model Loading Time: {(model_load_end - model_load_start) * 1000:.2f} ms")
print(f"Input Preparation Time: {(input_prep_end - input_prep_start) * 1000:.2f} ms")
print(f"Inference Time: {(inference_end - inference_start) * 1000:.2f} ms")
print(f"Total Time (including setup): {(overall_end_time - overall_start_time) * 1000:.2f} ms")
```

```python
predicted_class = output.argmax(dim=1).item()

print(predicted_class)  # Output: 1

# Mapping to label (optional):
class_mapping = {
    0: 'glioma_tumor',
    1: 'meningioma_tumor',
    2: 'no_tumor',
    3: 'pituitary_tumor'
}

print(class_mapping[predicted_class])  # Output: meningioma_tumor
```
